```python

# ================================================================
# PROJECT:
# AI-BASED INVENTORY DEMAND FORECASTING & STOCK OPTIMIZATION
#
# Dataset: inventory_dataset.csv
# Algorithm: Random Forest Regression
# Platform: Google Colab
# ================================================================


# ================================================================
# 1. INSTALL AND IMPORT LIBRARIES
# ================================================================

!pip install pandas numpy matplotlib seaborn scikit-learn -q

import os
import shutil
import warnings
import zipfile

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings('ignore')
sns.set_style('whitegrid')

# Create output folders

OUTPUT_DIR = 'Inventory_Project_Results'
GRAPH_DIR = os.path.join(OUTPUT_DIR, 'Graphs')

os.makedirs(GRAPH_DIR, exist_ok=True)

print("Libraries imported successfully!")


# ================================================================
# 2. UPLOAD YOUR REAL DATASET
# ================================================================

print("Please upload inventory_dataset.csv")

uploaded = files.upload()

if 'inventory_dataset.csv' not in uploaded:
    raise ValueError(
        "Please upload a file named inventory_dataset.csv"
    )

df = pd.read_csv('inventory_dataset.csv')

print("\nDataset uploaded successfully!")

print("\nFirst five records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# ================================================================
# 3. AUTOMATICALLY IDENTIFY DATASET COLUMNS
# ================================================================

# Make column matching independent of capitalization
column_lookup = {
    str(col).strip().lower().replace(' ', '_'): col
    for col in df.columns
}

def find_column(possible_names):
    for name in possible_names:
        if name in column_lookup:
            return column_lookup[name]
    return None


date_col = find_column([
    'date',
    'order_date',
    'sales_date',
    'transaction_date'
])

product_col = find_column([
    'product',
    'item',
    'sku',
    'product_id',
    'item_id'
])

demand_col = find_column([
    'demand',
    'sales',
    'quantity_sold',
    'units_sold',
    'order_quantity'
])

store_col = find_column([
    'store',
    'store_id',
    'location',
    'warehouse'
])

promotion_col = find_column([
    'promotion',
    'promo',
    'is_promotion',
    'discount'
])

stock_col = find_column([
    'current_stock',
    'stock_quantity',
    'quantity_on_hand',
    'inventory_level',
    'stock'
])

print("\nDetected Columns:")
print("Date:", date_col)
print("Product:", product_col)
print("Demand:", demand_col)
print("Store:", store_col)
print("Promotion:", promotion_col)
print("Current Stock:", stock_col)

if date_col is None or product_col is None or demand_col is None:
    raise ValueError(
        "Required columns not found. Your dataset must contain "
        "a date column, product/item column, and sales/demand column. "
        "Please check the printed column names and update the "
        "column mapping in Section 3."
    )


# ================================================================
# 4. DATA PREPROCESSING
# ================================================================

# Standardize the important columns

df['Date'] = pd.to_datetime(
    df[date_col],
    errors='coerce'
)

df['Product'] = df[product_col].astype(str)

df['Demand'] = pd.to_numeric(
    df[demand_col],
    errors='coerce'
)

# Promotion is optional

if promotion_col is not None:
    df['Promotion'] = pd.to_numeric(
        df[promotion_col],
        errors='coerce'
    ).fillna(0)
else:
    df['Promotion'] = 0

# Current stock is optional

if stock_col is not None:
    df['Current_Stock'] = pd.to_numeric(
        df[stock_col],
        errors='coerce'
    )

# Remove invalid date and demand records

df.dropna(
    subset=['Date', 'Product', 'Demand'],
    inplace=True
)

# Remove negative demand

df = df[df['Demand'] >= 0].copy()

# Remove duplicates

df.drop_duplicates(inplace=True)

# Create a separate series for each store-product combination

if store_col is not None:
    df['Series'] = (
        df[store_col].astype(str)
        + '_'
        + df['Product']
    )
else:
    df['Series'] = df['Product']

# Extract date features

df['Day'] = df['Date'].dt.day
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df['DayOfWeek'] = df['Date'].dt.dayofweek

# Sort dataset

df.sort_values(
    ['Series', 'Date'],
    inplace=True
)

print("\nPreprocessing Completed!")

print("\nClean Dataset:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nDataset Shape:", df.shape)


# ================================================================
# 5. EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

print("\n========== EXPLORATORY DATA ANALYSIS ==========")

print("\nDataset Description:")
print(df.describe(include='all'))

# Numerical columns

numeric_df = df.select_dtypes(include=np.number)

print("\nNumerical Columns:")
print(numeric_df.columns.tolist())

print("\nFive Number Summary:")
print(numeric_df.describe().loc[
    ['min', '25%', '50%', '75%', 'max']
])


# ================================================================
# 6. OUTLIER DETECTION USING IQR
# ================================================================

print("\n========== OUTLIER DETECTION ==========")

outlier_results = []

for col in numeric_df.columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df[col] < lower) |
        (df[col] > upper)
    ).sum()

    outlier_results.append([
        col, lower, upper, count
    ])

    print(col, "Outliers:", count)

outlier_df = pd.DataFrame(
    outlier_results,
    columns=[
        'Column',
        'Lower_Bound',
        'Upper_Bound',
        'Outlier_Count'
    ]
)

outlier_df.to_csv(
    os.path.join(OUTPUT_DIR, 'outlier_analysis.csv'),
    index=False
)


# ================================================================
# 7. GRAPH 1: HISTOGRAM OF DEMAND
# ================================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    df['Demand'],
    kde=True,
    bins=30
)

plt.title('Distribution of Product Demand')
plt.xlabel('Demand')
plt.ylabel('Frequency')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '1_demand_histogram.png'),
    dpi=300
)

plt.show()


# ================================================================
# 8. GRAPH 2: AVERAGE DEMAND BY PRODUCT
# ================================================================

product_demand = (
    df.groupby('Product')['Demand']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

product_demand.plot(kind='bar')

plt.title('Average Demand by Product')
plt.xlabel('Product')
plt.ylabel('Average Demand')

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '2_product_demand.png'),
    dpi=300
)

plt.show()


# ================================================================
# 9. GRAPH 3: DAILY DEMAND TREND
# ================================================================

daily_demand = df.groupby('Date')['Demand'].sum()

plt.figure(figsize=(12, 5))

plt.plot(daily_demand.index, daily_demand.values)

plt.title('Daily Inventory Demand Trend')
plt.xlabel('Date')
plt.ylabel('Total Demand')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '3_daily_demand.png'),
    dpi=300
)

plt.show()


# ================================================================
# 10. GRAPH 4: MONTHLY DEMAND ANALYSIS
# ================================================================

monthly_demand = (
    df.groupby(df['Date'].dt.month)['Demand']
    .sum()
)

plt.figure(figsize=(10, 5))

monthly_demand.plot(kind='bar')

plt.title('Monthly Demand Analysis')
plt.xlabel('Month')
plt.ylabel('Total Demand')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '4_monthly_demand.png'),
    dpi=300
)

plt.show()


# ================================================================
# 11. GRAPH 5: CORRELATION HEATMAP
# ================================================================

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '5_correlation_heatmap.png'),
    dpi=300
)

plt.show()


# ================================================================
# 12. GRAPH 6: SCATTER PLOT WITH TREND LINE
# ================================================================

# Use a sample for large datasets

scatter_df = df.sample(
    n=min(3000, len(df)),
    random_state=42
)

plt.figure(figsize=(10, 5))

sns.regplot(
    data=scatter_df,
    x='Day',
    y='Demand',
    scatter_kws={'alpha': 0.3}
)

plt.title('Day vs Demand Analysis')
plt.xlabel('Day of Month')
plt.ylabel('Demand')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '6_scatter_demand.png'),
    dpi=300
)

plt.show()

print("\nCorrelation between Day and Demand:")
print(df['Day'].corr(df['Demand']))


# ================================================================
# 13. GRAPH 7: OUTLIER BOXPLOT
# ================================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df['Demand']
)

plt.title('Demand Outlier Detection')
plt.xlabel('Demand')

plt.tight_layout()

plt.savefig(
    os.path.join(GRAPH_DIR, '7_demand_boxplot.png'),
    dpi=300
)

plt.show()


# ================================================================
# 14. FEATURE ENGINEERING
# ================================================================

df = df.sort_values(
    ['Series', 'Date']
).copy()

# Previous day demand

df['Lag_1'] = (
    df.groupby('Series')['Demand']
    .shift(1)
)

# Previous seven observations

df['Lag_7'] = (
    df.groupby('Series')['Demand']
    .shift(7)
)

# Previous seven-observation rolling average

df['Rolling_Mean_7'] = (
    df.groupby('Series')['Demand']
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)

# Remove rows without historical information

df.dropna(
    subset=[
        'Lag_1',
        'Lag_7',
        'Rolling_Mean_7'
    ],
    inplace=True
)

# Encode product/store series

df['Series_ID'] = pd.factorize(
    df['Series']
)[0]

features = [
    'Series_ID',
    'Promotion',
    'Day',
    'Month',
    'Year',
    'DayOfWeek',
    'Lag_1',
    'Lag_7',
    'Rolling_Mean_7'
]

X = df[features]
y = df['Demand']

print("\nFeature Engineering Completed!")
print("Features:", features)


# ================================================================
# 15. TRAIN-TEST SPLIT
# ================================================================

# Chronological split to avoid future information
# entering the training dataset

split_date = df['Date'].quantile(0.80)

train = df[
    df['Date'] <= split_date
].copy()

test = df[
    df['Date'] > split_date
].copy()

if len(train) == 0 or len(test) == 0:
    raise ValueError(
        "Not enough data for train-test split. "
        "Use a dataset with more historical records."
    )

X_train = train[features]
X_test = test[features]

y_train = train['Demand']
y_test = test['Demand']

print("\nTraining Records:", len(train))
print("Testing Records:", len(test))

print("Training End Date:", train['Date'].max())
print("Testing Start Date:", test['Date'].min())


# ================================================================
# 16. TRAIN RANDOM FOREST REGRESSION MODEL
# ================================================================

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    max_depth=12,
    n_jobs=-1
)

model.fit(
    X_train,
    y_train
)

print("\nRandom Forest Model Trained Successfully!")


# ================================================================
# 17. DEMAND FORECASTING
# ================================================================

predictions = model.predict(X_test)

predictions = np.maximum(
    predictions,
    0
)

results = test[
    ['Date', 'Product', 'Series', 'Demand']
].copy()

results['Predicted_Demand'] = predictions

results.to_csv(
    os.path.join(
        OUTPUT_DIR,
        'demand_forecast_results.csv'
    ),
    index=False
)

print("\nDemand Forecast Results:")
print(results.head(10))


# ================================================================
# 18. MODEL PERFORMANCE EVALUATION
# ================================================================

mae = mean_absolute_error(
    y_test,
    predictions
)

mse = mean_squared_error(
    y_test,
    predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    predictions
)

print("\n========== MODEL PERFORMANCE ==========")

print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))

performance_df = pd.DataFrame({
    'Metric': [
        'MAE',
        'MSE',
        'RMSE',
        'R2 Score'
    ],
    'Value': [
        mae,
        mse,
        rmse,
        r2
    ]
})

performance_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        'model_performance.csv'
    ),
    index=False
)


# ================================================================
# 19. GRAPH 8: ACTUAL VS PREDICTED DEMAND
# ================================================================

plot_results = results.sort_values('Date').head(200)

plt.figure(figsize=(14, 6))

plt.plot(
    plot_results['Date'],
    plot_results['Demand'],
    label='Actual Demand'
)

plt.plot(
    plot_results['Date'],
    plot_results['Predicted_Demand'],
    label='Predicted Demand'
)

plt.title('Actual vs Predicted Demand')
plt.xlabel('Date')
plt.ylabel('Demand')

plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        GRAPH_DIR,
        '8_actual_vs_predicted.png'
    ),
    dpi=300
)

plt.show()


# ================================================================
# 20. INVENTORY STOCK OPTIMIZATION
# ================================================================

# Inventory planning assumptions

lead_time = 7

planning_days = 30

z_score = 1.65

inventory_data = []

for series in results['Series'].unique():

    product_data = results[
        results['Series'] == series
    ]

    original_data = df[
        df['Series'] == series
    ]

    avg_predicted_demand = (
        product_data['Predicted_Demand'].mean()
    )

    # Historical demand variation
    demand_std = original_data['Demand'].std()

    if pd.isna(demand_std):
        demand_std = 0

    # Safety stock calculation
    safety_stock = (
        z_score * demand_std * np.sqrt(lead_time)
    )

    # Reorder point
    reorder_point = (
        avg_predicted_demand * lead_time
        + safety_stock
    )

    # Recommended stock for 30 days
    recommended_stock = (
        avg_predicted_demand * planning_days
        + safety_stock
    )

    # Current stock if available in original dataset
    current_stock = np.nan

    if stock_col is not None:
        stock_values = original_data[
            'Current_Stock'
        ].dropna()

        if len(stock_values) > 0:
            current_stock = stock_values.iloc[-1]

    inventory_data.append([
        series,
        avg_predicted_demand,
        demand_std,
        safety_stock,
        reorder_point,
        recommended_stock,
        current_stock
    ])

inventory_df = pd.DataFrame(
    inventory_data,
    columns=[
        'Product',
        'Predicted_Daily_Demand',
        'Historical_Demand_Std',
        'Safety_Stock',
        'Reorder_Point',
        'Recommended_Stock',
        'Current_Stock'
    ]
)

# Calculate order quantity only when current stock exists

if stock_col is not None:

    inventory_df['Stock_Status'] = np.where(
        inventory_df['Current_Stock']
        <= inventory_df['Reorder_Point'],
        'Reorder Required',
        'Stock Available'
    )

    inventory_df['Suggested_Order_Quantity'] = (
        np.maximum(
            0,
            np.ceil(
                inventory_df['Recommended_Stock']
                - inventory_df['Current_Stock']
            )
        )
    )

else:

    inventory_df['Stock_Status'] = (
        'Current stock data unavailable'
    )

    inventory_df['Suggested_Order_Quantity'] = np.nan

# Round numeric results

numeric_columns = [
    'Predicted_Daily_Demand',
    'Historical_Demand_Std',
    'Safety_Stock',
    'Reorder_Point',
    'Recommended_Stock'
]

inventory_df[numeric_columns] = (
    inventory_df[numeric_columns].round(2)
)

print("\n========== INVENTORY OPTIMIZATION ==========")

print(inventory_df.to_string(index=False))


# ================================================================
# 21. GRAPH 9: INVENTORY STOCK OPTIMIZATION
# ================================================================

plt.figure(figsize=(14, 6))

inventory_df.set_index('Product')[
    [
        'Safety_Stock',
        'Reorder_Point',
        'Recommended_Stock'
    ]
].plot(
    kind='bar',
    figsize=(14, 6)
)

plt.title('Inventory Stock Optimization')
plt.xlabel('Products')
plt.ylabel('Stock Quantity')

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        GRAPH_DIR,
        '9_stock_optimization.png'
    ),
    dpi=300
)

plt.show()


# ================================================================
# 22. GRAPH 10: PRODUCT-WISE PREDICTED DEMAND
# ================================================================

predicted_product_demand = (
    results.groupby('Product')['Predicted_Demand']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

predicted_product_demand.plot(
    kind='bar'
)

plt.title('Average Predicted Demand by Product')
plt.xlabel('Product')
plt.ylabel('Predicted Demand')

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    os.path.join(
        GRAPH_DIR,
        '10_predicted_product_demand.png'
    ),
    dpi=300
)

plt.show()


# ================================================================
# 23. SAVE INVENTORY OPTIMIZATION RESULTS
# ================================================================

inventory_df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        'inventory_optimization_results.csv'
    ),
    index=False
)

# Save cleaned dataset

df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        'cleaned_inventory_dataset.csv'
    ),
    index=False
)

print("\nAll CSV files saved successfully!")


# ================================================================
# 24. AUTOMATIC PROJECT SUMMARY
# ================================================================

print("\n========================================")
print("          PROJECT SUMMARY")
print("========================================")

print("Project: AI-Based Inventory Demand Forecasting")
print("Algorithm: Random Forest Regression")

print("Total Products/Series:",
      inventory_df['Product'].nunique())

print("Training Records:", len(train))
print("Testing Records:", len(test))

print("\nModel Performance:")

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))

print("\nInventory Recommendations:")

print(
    inventory_df[
        [
            'Product',
            'Predicted_Daily_Demand',
            'Safety_Stock',
            'Reorder_Point',
            'Recommended_Stock',
            'Stock_Status'
        ]
    ].to_string(index=False)
)

print("\nPROJECT COMPLETED SUCCESSFULLY!")


# ================================================================
# 25. DOWNLOAD CSV FILES INDIVIDUALLY
# ================================================================

print("\nDownloading individual CSV files...")

csv_files = [
    'inventory_optimization_results.csv',
    'demand_forecast_results.csv',
    'model_performance.csv',
    'cleaned_inventory_dataset.csv',
    'outlier_analysis.csv'
]

for filename in csv_files:

    path = os.path.join(
        OUTPUT_DIR,
        filename
    )

    files.download(path)


# ================================================================
# 26. CREATE ZIP FILE CONTAINING ALL RESULTS
# ================================================================

shutil.make_archive(
    'Inventory_Project_Complete_Results',
    'zip',
    OUTPUT_DIR
)

print("\nZIP file created successfully!")

files.download(
    'Inventory_Project_Complete_Results.zip'
)


# ================================================================
# 27. OPTIONAL: DOWNLOAD ALL GRAPHS INDIVIDUALLY
# ================================================================

print("\nDownloading all graphs...")

for filename in sorted(os.listdir(GRAPH_DIR)):

    if filename.endswith('.png'):

        files.download(
            os.path.join(GRAPH_DIR, filename)
        )

print("\n========================================")
print(" ALL PROJECT FILES GENERATED SUCCESSFULLY")
print("========================================")
```
